# Exploratory Data Analysis - E-commerce Sales

Quick look at trend, weekly seasonality, yearly seasonality, and promotion effects before modeling.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

df = pd.read_csv("../data/sales_data.csv", parse_dates=["date"])
df.head()
```

## 1. Overall sales trend over time

```python
plt.figure(figsize=(14, 5))
plt.plot(df["date"], df["sales"], linewidth=0.8)
plt.title("Daily Sales - Full History")
plt.xlabel("Date")
plt.ylabel("Sales ($)")
plt.show()
```

## 2. Weekly seasonality (average sales by day of week)

```python
df["weekday"] = df["date"].dt.day_name()
order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
weekday_avg = df.groupby("weekday")["sales"].mean().reindex(order)

plt.figure(figsize=(9, 5))
sns.barplot(x=weekday_avg.index, y=weekday_avg.values, palette="Blues_d")
plt.title("Average Sales by Day of Week")
plt.ylabel("Avg Sales ($)")
plt.show()
```

## 3. Monthly / yearly seasonality

```python
df["month"] = df["date"].dt.month
monthly_avg = df.groupby("month")["sales"].mean()

plt.figure(figsize=(9, 5))
sns.barplot(x=monthly_avg.index, y=monthly_avg.values, palette="Oranges_d")
plt.title("Average Sales by Month")
plt.xlabel("Month")
plt.ylabel("Avg Sales ($)")
plt.show()
```

## 4. Promotion impact

```python
promo_avg = df.groupby("is_promotion")["sales"].mean()
print(promo_avg)

plt.figure(figsize=(6, 5))
labels = promo_avg.index.map({0: "No Promotion", 1: "Promotion"})
sns.barplot(x=labels, y=promo_avg.values, palette="Greens_d")
plt.title("Average Sales: Promotion vs. No Promotion")
plt.ylabel("Avg Sales ($)")
plt.show()
```

## 5. Rolling trend (30-day moving average)

```python
df["rolling_30"] = df["sales"].rolling(30).mean()

plt.figure(figsize=(14, 5))
plt.plot(df["date"], df["sales"], alpha=0.3, label="Daily Sales")
plt.plot(df["date"], df["rolling_30"], color="red", linewidth=2, label="30-Day Rolling Avg")
plt.legend()
plt.title("Sales with 30-Day Rolling Average (Trend)")
plt.show()
```
